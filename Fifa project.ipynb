### * Filling Missing Values

```python
df.isnull().sum()
```

 - to view all the column we use display max rows as below

```python
null_values = df.isnull().sum()
pd.set_option('display.max_rows', None)
null_values
```

```python
# Identify object columns

object_columns = df.select_dtypes(include=['object']).columns

# Fill missing values with the mode of the column

for column in object_columns:
    mode_value = df[column].mode()[0]
    df[column] = df[column].fillna(mode_value)
```

```python
# Identify numerical columns

numerical_columns = df.select_dtypes(include=['number']).columns

# Fill missing values with the mean of the column

for column in numerical_columns:
    mean_value = df[column].mean()
    df[column] = df[column].fillna(mean_value)
```

#### 2 . Plot the distribution of overall rating vs. age of players. Interpret what is the age after which a player stops improving?

```python
# Plot the distribution of overall rating vs. age of players

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(5, 3)) 
sns.scatterplot(x=df['age'], y=df['overall'], color = 'orange') 
plt.title('Overall Rating vs. Age')
plt.xlabel('Age')
plt.ylabel('Overall Rating')
plt.show()
```

```python
# Interpret what is the age after which a player stops improving

import statsmodels.api as sm
plt.figure(figsize=(5, 3))

# Fit LOESS trend line

lowess = sm.nonparametric.lowess
smoothed = lowess(df['overall'], df['age'])

# Plot the trend line

plt.plot(smoothed[:, 0], smoothed[:, 1], color = 'purple')
```

```python
# Ploting the distribution of overall rating vs. age of players & Interpret what is the age after which a player stops improving ?

# Combining both the scatter plot and Fit LOESS trend line 

plt.figure(figsize=(6, 4))
sns.scatterplot(x=df['age'], y=df['overall'], color = 'orange')
plt.title('Overall Rating vs. Age')
plt.xlabel('Age')
plt.ylabel('Overall Rating')

import statsmodels.api as sm

# Fit LOESS trend line

lowess = sm.nonparametric.lowess
smoothed = lowess(df['overall'], df['age'])

# Plot the trend line

plt.plot(smoothed[:, 0], smoothed[:, 1], color='purple')
```

 - As we observe in the above plot, after reaching at a point the trend line has become Stagnant which mean,after that point the player preformance has stoped imporving.
 - Trend Line Analysis:

   -  The trend line shows an upward trajectory up to around age 25, indicating that players' overall ratings tend to   increase with age up to this point.

   -  After age 25, the trend line levels off, suggesting a plateau in the improvement of overall ratings.

   -  Beyond age 30, the trend line gradually declines, indicating a decrease in overall ratings.

#### 3 . Which type of offensive players tends to get paid the most: the striker, the right-winger, or the left-winger?

```python
# Filter players by offensive positions

offensive_positions = ['ST', 'RW', 'LW']
offensive_players = df[df['player_positions'].isin(offensive_positions)]

# Calculate average wage for each offensive position

avg_wages = offensive_players.groupby('player_positions')['wage_eur'].mean()

# Print the results

print("Average wage by offensive position:")
print(avg_wages)
```

 - from the above output we can see that among the average wage of st, lw, rw the 'st' that is 'striker' is paid high salary or wage.

### 1 . K-Means Clustering

```python
from sklearn.cluster import KMeans

# Initialize the KMeans model with 2 clusters
model = KMeans(n_clusters=2)

# Fit the model to the DataFrame 'df'
model.fit(df)

# Retrieve the coordinates of the cluster centers
model.cluster_centers_

# Retrieve the labels assigned to each data point
model.labels_
```

```python
# Calculate the silhouette score to evaluate the clustering quality

from sklearn.metrics import silhouette_score
silhouette_score(df, model.labels_)
```

#### * Elbow Technique

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial.distance import cdist

error = []
for i in range (1,71) :
    model = KMeans(n_clusters = i)
    model.fit(df)
    dist = cdist(df, model.cluster_centers_)
    act_dist = np.min(dist, axis = 1)
    total_error = sum(act_dist)
    error.append(total_error)

plt.plot(range(1,71), error)
plt.show()
```
